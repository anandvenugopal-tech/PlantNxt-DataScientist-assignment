# Data Cleaning & Preprocessing
This notebook focuses on cleaning and preprocessing the dataset to improve data quality and prepare it for further analysis and forecasting tasks.

Import libraries

```python
import numpy as np
import pandas as pd
```

Load & inspect data

```python
df = pd.read_csv(r'C:\PlantNxt Assignment\data\raw\revenue_ledger.csv')
df.head()
```

```python
df.tail()
```

```python
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

Datatype conversion

```python
columns = ['Price', 'Tax', 'Others']

for col in columns:
    df[col] = df[col].astype(str).str.replace(',', '').str.strip()
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['Inv Date'] = pd.to_datetime(df['Inv Date'])
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
```

```python
df.dtypes
```

Null value Handling

```python
df.isna().sum()
```

```python
(df.isna().sum()/len(df)) * 100
```

```python
df['Others'] = df['Others'].fillna(df['Others'].mean())
df['Tax'] = df['Tax'].fillna(df['Tax'].mean())
```

Handling duplicates

```python
print(f"No: of duplicates: {df.duplicated().sum()}")
print(f"% of duplicates: {df.duplicated().sum()/len(df) * 100: .2f}%")
```

```python
df[df.duplicated()].head()
```

```python
df = df.drop_duplicates()
df = df.reset_index(drop=True)
df.shape
```

save the cleaned data

```python
df.to_csv(r'C:\PlantNxt Assignment\data\processed\data.csv', index = False)
```

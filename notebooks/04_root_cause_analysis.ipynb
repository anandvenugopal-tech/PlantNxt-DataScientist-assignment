### Root Cause & What-If Analysis

Observation: Revenue in 2025 shows a massive spike compared to previous years. Annual revenue increased from around ₹1.8B in earlier years to more than ₹5.5B in 2025. Here we are analysing what are the root cause for this observation


Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Load data

```python
df = pd.read_csv('C:\PlantNxt Assignment\data\processed\data.csv')
```

Add relevant features

```python
df['Inv Date'] = pd.to_datetime(df['Inv Date'])
df['Revenue'] = df['Quantity'] * df['Price']
df["Year"] = df['Inv Date'].dt.year
df['Month_Year'] = df['Inv Date'].dt.to_period('M')
```

Yearly Trend 

```python
yearly = df.groupby('Year')['Revenue'].sum()

plt.figure(figsize=(14, 5))
plt.bar(yearly.index, yearly.values)
plt.title('Revenue by Year (2022 -2026)')
plt.xlabel('Year')
plt.ylabel('Revenue (INR)')
plt.show()
```

Monthly Trend

```python
monthly_rev = df.groupby('Month_Year')['Revenue'].sum()
monthly_rev.index = monthly_rev.index.to_timestamp()

plt.figure(figsize=(12, 6))
plt.plot(monthly_rev.index, monthly_rev.values, marker='o', color='green', linewidth=2)
plt.axvline(pd.Timestamp('2025-01-01'), color='red', linestyle='--')

plt.title('Monthly Revenue Performance (2022-2026)', fontsize=14)
plt.ylabel('Revenue (INR Millions)')
plt.xlabel('Date')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Root Cause Analysis

```python
yearly_revenue = df.groupby(['Year', 'Account'])['Revenue'].sum().unstack(fill_value=0)


yearly_revenue.plot(kind='bar', stacked=True, figsize=(10, 6), color=['#1f77b4', '#2ca02c'])

plt.title('Proof of Growth: Account B Entry Impact', fontsize=14)
plt.ylabel('Total Revenue (INR)', fontsize=12)
plt.xlabel('Year', fontsize=12)
plt.legend(title='Account')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add numeric totals on top of bars
for i, total in enumerate(yearly_revenue.sum(axis=1)):
    plt.text(i, total + 5e7, f'₹{total/1e9:.1f}B', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()


```

### The Root Cause
Revenue Growth Percentage: The data proves a 215.1% growth in total revenue between 2024 and 2025.

The "Before" (2022–2024): The blue bars (Account A) show a stagnant business hovering around ₹1.7B – ₹1.9B annually.

The "After" (2025–Present): The introduction of the green bar (Account B) in 2025 instantly added ₹3.6 Billion in new revenue—more than doubling the size of the entire company in a single year.

2026 Trajectory: Even with only partial data for 2026, Account B has already contributed ₹1.83 Billion, surpassing Account A’s total for almost any previous full year.


### What if no Account B

1. The "Scale" Impact (Size)
   Without Account B: Your total lifetime revenue would drop from ₹13.31 Billion to just ₹7.85 Billion.

   The Interpretation: The company would be 41% smaller. You would effectively lose nearly half of your business volume, likely leading to massive under-utilization of your factory machines and staff.

2. The "Growth" Impact (Velocity)
   Actual (With B): The business grew by 215% in 2025.

   What-If (No B): Revenue would have remained flat or stagnant at around ₹1.7B – ₹1.9B per year, as it was from 2022 to 2024.

   The Interpretation: Without Account B, the company would have zero momentum. It would be a "survival" business rather than a "growth" business.
